```python
#Importing the libraries
import numpy as np
import pandas as pd
```

```python
#Reading the csv file and loading the data.
data=pd.read_csv("indian_realestate_linear.csv")
data.head()
```

```python
#Data Cleaning.
#Checking the total information of the data.

data.info()
```

```python
#Checking the shape of a data.

data.shape
```

```python
#Checking the total no of columns present in the data.

data.columns
```

```python
#Checking the describe method to get the statistical analysis for the data.

data.describe()
```

```python
#Verifying the any null values present in the data.

data.isnull().sum()
```

```python
#Removing the unwanted data

data1=data.drop(['property_id','property_type','price_per_sqft_inr','facing','amenities','amenities_count',
       'walkability_score','monthly_rental_estimate_inr',
       'annual_rental_yield_pct','price_category','estimated_roi_5yr_pct',
       'listing_date','days_on_market','status','registration_days',
       'is_gated_community','has_maintenance_fee','monthly_maintenance_inr'],axis=1)
data1
```

```python
# Here city ,district is in categorical columns .
#By converting it into numerical columns using onehot encoder because it works on multiple columns.
#Importing onehot encoder

from sklearn.preprocessing import OneHotEncoder
categorical_cols = ["city", "district"]
ohe=OneHotEncoder(sparse_output=False).set_output(transform ='pandas')
ohe_encoded=ohe.fit_transform( data1[["city", "district"]])
ohe_encoded
```

```python
##Combining the features.

data2=pd.concat([data1,ohe_encoded],axis=1).drop(["city","district"],axis=1)
data2.head()
```

```python
#Applying a log transformation to the 'price_inr' column.
# This transformation reduces skewness in the data when price vary widely.

data2["log_price"] = np.log1p(data2["price_inr"])

```

```python
#showing the independent and target features.
#Independent Variable.

x=data2.drop(["price_inr"],axis=1)

#Target Variable.
y=data2["price_inr"]
```

```python
data2.head()
```

```python
#showing the correlation for target and independent variable.

import matplotlib.pyplot as plt
import seaborn as sns

#Setting the figure size
plt.figure(figsize=(12,10))

#Heatmap plot
corr_matrix = data2.corr(numeric_only=True)
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")

# Adding the title

plt.title("Correlation Heatmap of Real Estate Dataset", fontsize=12)

#Displaying the plot
plt.show()
```

```python
#Using Train-test-split to train the data.

from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# Shape of the model

print("Train_shape:",x_train.shape,y_train.shape)
print("Test_shape:",x_test.shape,y_test.shape)
```

```python
#Fitting data into the model.
#importing the libraries 

from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
#Predicting on test set.
y_pred=model.predict(x_test)
y_pred
```

```python
#Showing the coefficients and intercept for the model.

print("coefficients:",model.coef_)
print("Intercept:",model.intercept_)
```

```python
#Evaluating the model by calculating the mean_absolute error,mean_squared_error,r2_score.

from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
print("Test_MAE:",mean_absolute_error(y_test,y_pred))
print("Test_MSE:",mean_squared_error(y_test,y_pred))
print("Test_r2_score:",r2_score(y_test,y_pred))
```

```python
#Visualizing the regression performance.

# Scatter plot of actual vs predicted values
plt.scatter(y_test, y_pred)  

# This line shows where predictions would exactly equal actual values
plt.plot([y_test.min(), y_test.max()], 
         [y_test.min(), y_test.max()], 
         "r--", label="Perfect_fit")  

# Adding label to the plot
plt.xlabel("Actual Values")  
plt.ylabel("Predicted Values")  

# Adding title to the plot
plt.title("Regression Performance Visualization")  

# Showing the legend (so the red dashed line is labeled)
plt.legend()  

# Displaying the plot
plt.show()
```
